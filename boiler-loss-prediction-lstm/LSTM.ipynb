```python
from sklearn.preprocessing import StandardScaler

features = data[['BOILER MASTER', 'HP FEEDWATER FLOW', 'MAIN STEAM FLOW', 'ECONOMIZER OUTLET TEMPERATURE', 'ECONOMIZER INLET TEMPERATURE', 'MAIN STEAM PRESSURE', 'MAIN STEAM TEMPERATURE', 'MAIN STEAM FLOW.1', 'AIR INLET APH TEMPERATURE', 'AIR OUTLET APH TEMPERATURE', 'AIR HEATER NB2 INLET TEMPERATURE', 'AIR HEATER NB1 INLET TEMPERATURE', 'TOTAL AIR FLOW ACTUAL', 'BOILER BANK GAS OUTLET BOILER 11', 'BOILER BANK GAS OUTLET TEMPERATOR', 'WET SCRUBBER GAS INLET TEMPEATOR', 'FLUE GAS ECONOMISER 1 INLET TEMPERATURE', 'FLUE GAS ECONOMISER 2 INLET TEMPERATURE', 'FLUE GAS ECONOMISER 2 OUTLET TEMPERATURE', 'FLUE GAS AH COOL OUTLET TEMPERATURE', 'STACK TEMPERATOR', 'FURNACE PRESSURE BOILER 11', 'BOILER 11 ID FAN CURRENT', 'ID FAN NDE BEARING TEMP', 'ID FAN DE BEARING TEMP', 'BOILER 11 ID FAN DAMPER']]
scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)
```

```python
from sklearn.decomposition import PCA
# กำหนดให้เก็บ variance 95%
pca = PCA(n_components=0.95)
principal_components = pca.fit_transform(features_scaled)
```

```python
import matplotlib.pyplot as plt
plt.plot(range(1, pca.n_components_+1), pca.explained_variance_ratio_, marker='o')
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance Ratio by Principal Component')
plt.show()
```

```python
# Define the target variables
target_cols = ['BOILER BANK GAS OUTLET OXYGEN', 'MAIN STEAM PRESSURE']
y = data[target_cols]

principal_components_selected = principal_components[:, [0, 1, 3, 4 ,6,12,13 , 8]]
# Define the training size (e.g., 80% of the data for training)
train_size = int(len(principal_components) * 0.8)
```

```python
# Split the data into training and testing sets
X_train = principal_components_selected[:train_size]
X_test = principal_components_selected[train_size:]
y_train = y[:train_size]
y_test = y[train_size:]
```

```python
def create_sequences(data, target, timesteps=40):
    X, y = [], []
    for i in range(len(data)-timesteps):
        X.append(data[i:i+timesteps])
        y.append(target[i+timesteps])
    return np.array(X), np.array(y)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(principal_components_selected, y)

# Coefficients
coef_df = pd.DataFrame(model.coef_, columns=[f'PC{i+1}' for i in range(8)], index=target_cols)
print(coef_df)
```

```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (Input, Dense, LSTM, Dropout,
                                      BatchNormalization, Bidirectional)
import tensorflow as tf
import numpy as np
import joblib

# Hyperparameters
timesteps = 50
num_features = X_train.shape[1]
lstm_units_1 = 128
lstm_units_2 = 64
dropout_rate = 0.2

# Input
inputs = Input(shape=(timesteps, num_features))

# LSTM Layer 1 - Bidirectional
x = Bidirectional(LSTM(lstm_units_1, return_sequences=True))(inputs)
x = Dropout(dropout_rate)(x)
x = BatchNormalization()(x)

# LSTM Layer 2
x = LSTM(lstm_units_2, return_sequences=False)(x)
x = Dropout(dropout_rate)(x)
x = BatchNormalization()(x)

# Dense layers
x = Dense(64, activation='relu')(x)
x = Dropout(dropout_rate)(x)

# Output layer (2 outputs)
outputs = Dense(2)(x)

model = Model(inputs=inputs, outputs=outputs)
model.compile(optimizer='adam', loss='mse', metrics=['mae'])

y_train_np = y_train.values
y_test_np = y_test.values

X_train_seq, y_train_seq = create_sequences(X_train, y_train_np, timesteps)
X_test_seq, y_test_seq = create_sequences(X_test, y_test_np, timesteps)

model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# Callback: หยุด training ถ้า validation loss ไม่ดีขึ้น 10 epochs
early_stop = EarlyStopping(
    monitor='val_loss',      # ตรวจสอบ validation loss
    patience=50,             # ถ้าไม่ดีขึ้น 10 epochs → หยุด
    restore_best_weights=True # โหลด weights ที่ดีที่สุดกลับมา
)

# Callback: บันทึก model ที่ validation loss ต่ำสุด
checkpoint = ModelCheckpoint(
    'best_lstm_model.h5',   # ไฟล์ที่จะบันทึก
    monitor='val_loss',
    save_best_only=True,
    verbose=1
)

# Train model พร้อม callbacks
history = model.fit(
    X_train_seq, y_train_seq,
    epochs=150,               # ปรับจำนวน epochs ตามต้องการ
    batch_size=64,
    validation_split=0.2,
    callbacks=[early_stop, checkpoint]
)
```

```python
import matplotlib.pyplot as plt

# วาดกราฟ Loss
plt.figure(figsize=(10,5))
plt.plot(history.history['loss'], label='Training Loss', marker='o')
plt.plot(history.history['val_loss'], label='Validation Loss', marker='o')
plt.title('Model Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True)
plt.show()

# วาดกราฟ MAE (ถ้า metrics=['mae'] ด้วย)
if 'mae' in history.history:
    plt.figure(figsize=(10,5))
    plt.plot(history.history['mae'], label='Training MAE', marker='o')
    plt.plot(history.history['val_mae'], label='Validation MAE', marker='o')
    plt.title('Model MAE Over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Mean Absolute Error')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
from sklearn.metrics import mean_squared_error, r2_score
y_pred = model.predict(X_test_seq)
print("MSE:", mean_squared_error(y_test_seq, y_pred))
print("R2:", r2_score(y_test_seq, y_pred))
```

```python
smape = 100/len(y_test_seq) * np.sum(2 * np.abs(y_pred - y_test_seq) / (np.abs(y_pred) + np.abs(y_test_seq)))
print("SMAPE:", smape, "%")
```
